```python
import sys
from itertools import product
from pathlib import Path

from core import (
  grid_tuning,
  load_data,
)
from strats.dema import f0_be_exit

sys.path.insert(0, str(Path.cwd().parent))

# CONFIG
TUNING = {
  "ASSETS": ["US500_scaled"],
  "TIMEFRAMES": ["M5", "M15"]
}
TUNING_GRID = list(product(*(v for v in TUNING.values())))


COMMISSION = 0.00015
CASH = 50_000

RISK_PER_TRADE = 10
RISK_PCT = RISK_PER_TRADE / CASH 

strat = f0_be_exit.STRAT(
   CASH=CASH,
   COMMISSION=COMMISSION,
   RISK_PER_TRADE=RISK_PER_TRADE,
)

for asset, tf in TUNING_GRID:
  name = f"{asset}_{tf}"
  TABLE_NAME = f'{strat.CODE}_{name}'
  DATA_PATH = f'~/market_data/{asset}/{name}.csv'

  ohlcv_data = load_data(DATA_PATH)
  grid_tuning(params=strat.PARAMS, table_name=TABLE_NAME, data=ohlcv_data, run_backtest=strat.run_backtest)
```
